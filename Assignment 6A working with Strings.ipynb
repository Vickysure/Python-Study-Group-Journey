## Ex3: Movie Recommendation Search Engine (Entertainment & AI Industry)

```python
# the code below is to store a movie title in a variable called 'movie'
movie = 'The Last Kingdom Returns'

# the code below is to display the first character in the movie title
# to get the first character, we use the slicing index method on the variable 'movie'
print (f'''First Character:
{movie[0:1]}
''')

# the code below is to display the last character in the movie title
# to get the last character, we use the slicing method 
print (f'''Last Character:
{movie[23:24]}
''')

# the code below is to display the first word in the movie title
# to get the first word, we use the slicing method
print (f'''First Word:
{movie[0:3]}
''')

# the code below is to display the last word in the movie title
# to get the last word, we use the slicing method
print (f'''Last Word:
{movie[17:24]}
''')

# the code below is to display the position of the word 'Returns' in the movie title
# to get the position of the word 'Returns', we use the [.find] method on the movie title
print (f'''Position of the Word 'Returns':
{movie.find('Returns')}
''')

# the code below is to split the words in the movie title
# to split the words in the movie title, we use the .split method on the movie title
print (f'''Split Result:
{movie.split()}
''')

```
