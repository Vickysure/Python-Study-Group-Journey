# Mastering String Slicing

## Ex1: Employee Record Information Extractor (Human Resources)

```python
# th code below is to create a variable for an employee record
employee_record = input ('Employee Record:',)

# the code below is to display the employee prefix from the employee record
# to display the employee prefix, we using the slicing index method and including the start and stop parameters
print (f'''Employee Prefix:
{employee_record[:3]}
''')

#the code below is to display the department code from the employee record
# to get the department code, we use the index method and including the start and stop parameters
print (f'''Department Code:
{employee_record [4:6]}
''')

# the code below is to display the employment year from the employee record
# to get the employment year, we use the index method and including the start and stop parameters
print (f'''Employment Year:
{employee_record[7:11]}
''')

# the code below is to display the employee number from the employee record
# to get the employee number, we use the index method and including the start and stop parameters
print (f'''Employee Number:
{employee_record[12:]}
''')

# the code below is to display the first ten characters of the employee record
# to get the first ten characters, we utilize the index method and including the start and stop parameters
print (f'''First Ten Characters:
{employee_record [:11]}
''')

# the code below is to display the last six characters of the employee record
# to get the last six characters, we use the index slicing method and including the start and stop parameters
print (f'''Last Six Characters:
{employee_record[12:]}
''')

# the code below is to display every second character from the employee record
# to get every second character of the employee record displayed, we use the index slicing method and including the start,stop and step parameters.
print (f'''Every Second Character:
{employee_record[1::2]}
''')

# the code below is to display every third character from the employee record
# to get every third character of the employee record displayed, we use the index slicing method and including the start,stop and step parameters.
print (f'''Every Third Character:
{employee_record[2::3]}
''')

# the code below is to display the entire record in reverse
# to get the entire reocrd in reverse, we use the index slicing method and including the start,stop and step parameters.
print (f'''Entire Record in Reverse:
{employee_record[::-1]}
''')
```

## Ex2: Banking Card Security Formatter (Banking & Finance)

```python
# the code below is to create a variable for a customer's card number
card_number = input ('Card Number:',)

# the code below is to display the first four digits using the string slicing method
print (f'''First Four Digits:
{card_number[:4]}
''')

# the code below is to display the last four digits using the string slicing method
print (f'''Last Four Digits:
{card_number[12:]}
''')

# the code below is to display the middle eight(8) digits using the string slicing method
print (f'''Middle Eight(8) Digits:
{card_number[4:12]}
''')

# the code below is to display every second digits using the string slicing method
print (f'''Every Second Digit:
{card_number[1::2]}
''')

# the code below is to display every third digit using the string slicing method
print (f'''Every Third Digit:
{card_number[2::3]}
''')

# the code below is to display the entire card number in reverse using the string slicing method
print (f'''Entire Card Number in Reverse:
{card_number[::-1]}
''')

# the code below is to display the last eight(8) digits in reverse order using the string slicing method
print (f'''Last Eight Digit in Reverse Order:
{card_number[15:7:-1]}
''')

#bonus challenge
# the code below is to display the last four digits in combination with masking characters using the slicing method only.
print ('****' +' '+'****'+' '+'****'+' '+(card_number[12:]))
```
