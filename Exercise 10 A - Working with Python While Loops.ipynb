## Ex 1: Transaction Processing Counter (Banking & Fintech)

```python
# Task:  the code below is to program an automated bank transaction processing counter using while loops

# Business Problem: the bank has indentified an issue regarding how effectively they can process customer transactions...
# ... as the number of customers transacting has started increasing daily, due to general acceptance of the bank by the public...
# ... now they can no longer process transactions manually, and therefore need an automated system to process the transactions...
# ... of millions of customers in the shortest time possible. So they decide to test their automated system by processs 10 transactions for a start.

# Solution Approach:
# i decided to solve this issue by using a while loop that would keep processing transactions as long as the transactions...
# ...for the day does not exceed a maximum of 10 transactions. 

# the variable below represents the initial transaction
transaction = 1

# the variable below represents the initial successful transaction
successful_transactions = 0

while transaction <= 10:
    print ('Processing Transaction', transaction)
    transaction +=1
    successful_transactions += 1
    
print (f'''
All 10 Transactions have been Processed Successfully''')
print (f'''
Sucessful Transactions: {successful_transactions}''')

# Expected vs Actual Outcome:
# the program ran successfully as expected and processed a total number of 10 transactions.

# Reflection:
# my loop control variable ; transaction <= 10, played an important role of letting the program know...
# ...it is expected to process only a number of 10 transactions.
# if the variable was not updated correctly, the program loop will enter an infinite loop.
```

### Transaction Processing Counter 2: Second Scenario.

```python

# the code below is for the program to processe 100 transactions, with an update of 10 transactions per batch.

# the variable below represents the initial transaction
transaction = 10

# the variable below represents the initial successful transaction
successful_transactions = 0

while transaction <= 100:
    print ('Processing Transaction', transaction)
    transaction +=10
    successful_transactions += 1
    
print (f'''
All 10 Transactions have been Processed Successfully''')
print (f'''
Sucessful Transactions: {successful_transactions}''')
```

## Ex2: Customer Input Validation System (Customer Service)

```python
# Business problem:
# An online company wants to automate their customer registration process under their customer service section...
# ... but only wants customers betweenn the ages of 18 and 60. Stating that any age number enter outside of the age range allowed...
# ... should be tagged invalid, and customers should have a maximum attempt of three trials when entering their age.

# Solution Approach:  
# I decided to use the while loop in the automation program, so that the sysem would keep requesting for the valid age...
#  ...once a user enters an age outside the accepted age range, until he or she enters the accpetable age.
# an update was later introduced to limit the attempts of entering their age to three (3), after which the system....
# ... should display 'Maximum Attempts Reached, Registration cannot continue', and then stops the customer from registering.

# the code below is to create an input validation system using the while loop.

# customer_age = age = int (input ('Enter Your Age:', ))
attempts = 0

while attempts < 4:
    age = int (input ('Enter Your Age:',))
    attempts += 1

    
    if 18 <= age <= 60:
        print ('Age Accepted, Registration can Continue')
        break

    else:
        print ('invalid age, Please enter an age between 18 and 60.')

print (f'''--------------------------------------------------------------------

Number of Attempts: {attempts}''')

# Expected vs Actual outcome:
# the program worked perfectly as envisioned, even after an update was made.

# Reflection:
# the while loop was used to tell the system the maximum number of attempts to accept from each customer.
# without the variable being updated correctly, there would be an infinite attempts availale to the customers which is not...
# ... what the online company executives want.
```

#### Input Validation System Update: Input Attempts limited to three.

```python
#task: the code below is to limit user input attempt to three times maximum.

attempts = 0

while attempts < 3:
    age = int(input("Enter your age: "))
    attempts += 1

    if 18 <= age <= 60:
        print("Age Accepted, Continue Registration")
        break
    else:
        print("Invalid Age, Please enter an age between 18 and 60.")

print(f"""-----------------------------------------------------------------

Total attempts: {attempts}
""")

if attempts == 3 and not (18 <= age <= 60):
    print(f"""Maximum attempts reached, Registration cannot Continue.
""")
```
