## Ex1: E-Commerce Product Combination Generator (E-commerce & Retail)

```python
# Task: a Python program that uses nested while loops to generate every possible combination of product size and color.

# Business Problem:
# the e-commerce company does not want to manually generate the color-size combinations for each product they have ...
#... as that would be exhausting and there could be errors in the manual generation, so the opted for automated process...
# ... to make the generating process of size-color combinations of their product efficient and error free.

# Solution Approach:
# I decided to use nested while loop to solve the problem so the system can have an way to cross-combine the product...
# ... size ans colors, and also have room for possible increment in the number of product sizes and colors without crashing the system.


# the variables below represent the provided data for product sizes and colors.
sizes = ['Small', 'Medium', 'Large']
colors = ['Black', 'White', 'Blue']

# [i] represents the index for the list 'sizes'
# [j] represents the index for list 'colors'

i = 0
j = 0

# len() gives us the total number of contents inside both lists of sizes and colors

while i < len (sizes):
    j = 0
    while j < len (colors):
        print (sizes [i], colors [j])
        j += 1
        
    print ('----------------------')
    i+=1

# the code below is to display the total number of product combinations
print(f'''
All Possible Product Combinations of Size and Color
has been Generated.
''')

# Explanation:
# the outer while loop is responsible for the number of times products sizes can be combined
# the inner while loop is responsible for the number of times product colors can be combined
# [i] represents the index for the list of product sizes
# [j] represents the index for the list of product colors
# [j[ needs to reset so that the outer while loop on its next iteration can start afresh from the first index of the colors' list.
# the loop eventually terminates once the outer loop has completely performed its iterations on each index of the sizez' list.

# Reflection:
# if the (j = 0) is removed, the loop would stop after the first iteration has been performed on the first index of the sizes' list...
#... meaning it would not be able to perform the combinations of the second index of product sizes as [j] would be = 3...
#... but then, the inner while loop condition only runs as long as [j] is less than 3, which is the number of the items in the colors' list. 
```

## Ex2: Banking Transaction Risk Checker (Banking & Financial Services)

```python
# Task: a Python program that uses a nested while loop to compare the customer records against the transaction records.

# Business Problem:
# the bank discovered that detecting risky or fraudulent transactions is difficult to do manually, so they chose to deploy...
# ... an automated system that will do the task while setting a transaction risk threshold.

# Solution Approach:
# I decided to use a nested while loop with a combination of if-else statement so the system would be able to process all transactions...
# ... and also be able to detect transactions above the risk threshold easily.

# the variables below represent the provided data for number of customers and the transactions carried out.
customers = ['C001', 'C002', 'C003']
transactions = [45000, 125000, 75000]

# the variable below is the limit placed on transactions, any transaction above the limit is placed under review.
risk_threshold = 100000

# the variables below represent the index for both list 'customers' and list 'transactions'.
# [i] represents the index for the list 'customers'
# [j] represents the index for list 'transactions'

i = 0
j = 0

while i < len (customers):
    j = 0
    while j < len (transactions):
        if transactions [j] < risk_threshold:
            print ('Customer:', customers[i], '|', 'Transaction:', transactions [j], '|', 'Status: BELOW LIMIT')

        else:
            transactions [j] > risk_threshold
            print ('Customer:', customers[i], '|', 'Transaction:', transactions [j], '|', 'Status: REVIEW')
        j += 1

    print('-' * 65)
    i += 1
```

## Ex3: Hospital Appointment Scheduling Matrix (HealthCare)

```python
# Task: a Python program using nested while loops to generate the scheduling combinations.

# Business Problem: 
# the hospital wants to generate the possible appointments schdules for their available doctors within four different time slots...
# ... but doesn't want to do this manually, so they decided to automate the process  to avoid errors and also give room for ...
#... easy modification in the nearest future, should in case their list of doctors increase or reduce.

# Solution Approach:
# I chose to use a nested while loop to solve this problem, so that the automated system can be able to generate their appointment schedules...
# ... correctly based on their available doctors within their chosen time slots, while also giving room for implementing changes ...
# ... that could be made by the hospital management; be it on the doctors available or their chosen time slots.


# the variables below represent the provided data for number of doctors available and the time slots to fix them.
doctors = ['Dr. Ahmed', 'Dr. Grace', 'Dr. Michael']
time_slots = ['9:00 AM', '11:00 AM', '2:00 PM', '4:00PM']

# the variables below represent the index for both list 'doctors' and list 'time_slots'.
# [i] represents the index for the list 'doctors'
# [j] represents the index for list 'time_slots'

i = 0
j = 0

while i < len(doctors):
    j = 0
    while j < len (time_slots):
        print (doctors[i], '--->', time_slots[j])
        j += 1

    print ('-' * 25)
    i += 1
```
