## Ex1: User Message Collector

```python
# the code below is a program for an automated user message collector for an Ai customer support chat system belonging to an online marketting company.

while True:
    message = input ('Enter Message or Q to quit:')

    if message == 'Q' or 'q':
        print ('Message Process Terminated by User.')
        break

    print (message)
    print ('-' * 50)
    
```

## Ex2: Simple command line support system

```python
# the code below is a program for an automated customer service system for an online vendor, to guide the users on the...
#... type of commands available and what render the services requested.

while True:
    command = input ('What do you need? :',)

    if command == 'Help' or command == 'help':
        print ('Available command: Help, Status, Quit.')
        print ('-' * 50)

    elif command == 'Status' or command =='status':
        print ('Your details are being process, please be patient.')
        print ('-' * 50)

    elif command == 'Quit' or command == 'quit':
        break

    else:
        print ('Unknown command!')
        print ('-' * 50)

print ('Service succesfully rendered, Have a nice day!')
```

#### Ex2 System Update (Study Buddy Challenge): Two extra commands Added.

```python
# the code below shows an update on the system with two extra commands of 'name' and 'version' added to the system commands.

while True:
    command = input ('What do you need? :',)

    if command == 'Help' or command == 'help':
        print ('Available command: Help, Status, Quit.')
        print ('-' * 50)

    elif command == 'Status' or command =='status':
        print ('Your details are being process, please be patient.')
        print ('-' * 50)

    elif command == 'Name' or command == 'name':
        print ('Python')
        print ('-' * 50)

    elif command == 'Version' or command == 'version':
        print ('Python Support System V1.0')
        print ('-' * 50)

    elif command == 'Quit' or command == 'quit':
        break

    else:
        print ('Unknown command!')
        print ('-' * 50)

print ('Service succesfully rendered, Have a nice day!')
```

## Ex3 : Simple Data entry System

```python
# the code below is a program to allow user to enter an amount for sales contnously or user decides to terminate
# ... process by entering 0 in the input section.

while True:
    amount = int (input('Enter Sales Amount or 0 to finish'))

    if amount == 0:
            print ('Data Entry Completed!')
            break

    print (f'Amount: #{amount}')
    print ('-' * 50)
```

#### Ex3 System Update (Study Buddy Challenge) : Total Sales amount Added

```python
# the code below shows an update on the system, where the total sales amount entered by user is calculated and displayed.

# the variable below 'total_sales' is introduced to be able to monitor and add correctly the sales amount entered by user.
total_sales = 0

while True:
    amount = int (input('Enter Sales Amount or 0 to finish'))

    if amount == 0:
            print ('Data Entry Completed!')
            break


    print (f'Amount: #{amount}')
    print ('-' * 50)
# to get the total_sales, the amount entered by user each time is being added to the total_sales
    total_sales += amount
    
print ('-' * 50)
print (f'Total Sales: #{total_sales}')
```
