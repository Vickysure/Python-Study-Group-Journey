## Ex1: Employee Salary Processing System (Human resources and Payroll)

```python
# the code below is to create a function to calculate the gross salary of an employee.
def calculate_gross_salary (basic_salary, housing_allowance, transport_allowance, other_allowance):
    gross_salary = basic_salary + housing_allowance + transport_allowance + other_allowance
    return gross_salary

# the code below is the display of the calculation of the gross_salary of three Employees
employee1 = calculate_gross_salary (2000,3000,4000,1000)
employee2 = calculate_gross_salary (4000,5000,6000,2000)
employee3 = calculate_gross_salary (10000,2000,3000,1500)

# the code below is to display the Gross salary of three Employees
print (f'''Gross Salary of Employees
{'-' * 37}''')
print (f'Employee 1 Gross Salary: #{employee1}')
print (f'Employee 2 Gross Salary: #{employee2}')
print (f'Employee 3 Gross Salary: #{employee3}')
```

#### Bonus Challenge : a Function for Net Salary

```python
# the code below is to create a function for an employee's net salary.
def calculate_net_salary (gross_salary,tax_deductions):
    net_salary = gross_salary - tax_deductions
    return net_salary
    
# the code below is the display of the calculation of the gross_salary of three Employees
Employee_1 = calculate_net_salary (10000,500)
Employee_2 = calculate_net_salary (17000,700)
Employee_3 = calculate_net_salary (16500,650)

# the code below is to display the net salary of three Employees
print (f'''Net Salary of Employees
{'-' * 37}''')
print (f'Employee 1 Net Salary: #{Employee_1}')
print (f'Employee 2 Net Salary: #{Employee_2}')
print (f'Employee 3 Net Salary: #{Employee_3}')
```

## Ex2: Hospital Patient Record Formatter (Healthcare/ Data Management)

```python
# the code below is to create a function that accepts a patient's name, formats it and display the formatted name.

def format_patient_name (name):
    name = input ('Patient\' Name:')
    name = name.strip().title()
    print (f'Formatted Name: {name}')
    print ('-' * 40)

print('Patient 1')
format_patient_name('name')

print ('Pateint 2')
format_patient_name ('name')

print ('Patient 3')
format_patient_name ('name')

print ('Patient 4')
format_patient_name ('name')

print ('Patient 5')
format_patient_name ('name')
```

#### Bonus Challenge(System Update) : A case of an Empty Name

```python
# the code below shows an update to the function created earlier for a scenario where an empty space is entered on the..
# ... input section.
def format_patient_name (name):
    attempts = 0
    while attempts < 10:
        name = input ('Patient\' Name:')
        if name == '':
            print ('Patient Name cannot be Empty')
            attempts += 1
        else:
            name = name.strip().title()
            print (f'Formatted Name:{name}')
            print ('-' * 40)
            break

        if attempts == 10 and name == '':
            print ('Maximum Attempts Reached!, Please Try Again Later.')
            break

            
print('Patient 1')
format_patient_name('name')

print ('Pateint 2')
format_patient_name ('name')

print ('Patient 3')
format_patient_name ('name')
```

## Ex3: E-Commerce Order Price Calculator (E-commerce/ Retail)

```python
# the code below is to create a function that calculates the total order of a customer in a e-commerce application.

def calculate_order_total ():
    product_price = float (input ('product price:'))
    quantity = int (input ('quantity:'))
    sub_total = product_price * quantity
    discount_percentage = int (input('discount percentage'))
    discount = (sub_total /100) * discount_percentage
    final_price = sub_total - discount
    print (f'Total price: #{final_price:.2f}')
    print ('-' * 40)
    print (f'Thank You for Patronizing Us!')
    print (f'''{'-' * 40}
          ''')

# the codes below are to calculate five different orders with the above created function.

print ('Order 1')
calculate_order_total ()

print ('Order 2')
calculate_order_total ()

print ('Order 3')
calculate_order_total ()

print ('Order 4')
calculate_order_total ()

print ('Order 5')
calculate_order_total ()
```
