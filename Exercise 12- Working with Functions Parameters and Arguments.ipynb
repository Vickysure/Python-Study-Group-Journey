#### Function 3: Calculate Account Status

```python
# the code below is to determine a customer's status based on their total amount spent.

def calculate_account_status ():
    total_spent = float (input ('Total Spent:'))
    if total_spent >= 100000:
        loyalty_points = total_spent / 100
        print ('Loyalty Points:',loyalty_points)
        print ('Customer Status: Premium')
        print ('-' * 40)
        print (f'''Thank You For Your Patronage!
We hope to See You Again.''')
        print ('-' * 40)
        print ('')

    elif 50000 <= total_spent <= 99999:
        loyalty_points = total_spent / 100
        print ('Loyalty Points:',loyalty_points)
        print ('Customer Status: Gold')
        print ('-' * 40)
        print (f'''Thank You For Your Patronage!
We hope to See You Again.''')
        print ('-' * 40)
        print ('')


    elif 20000 <= total_spent <= 49999:
        loyalty_points = total_spent / 100
        print ('Loyalty Points:',loyalty_points)
        print ('Customer Status: Silver')
        print ('-' * 40)
        print (f'''Thank You For Your Patronage!
We hope to See You Again.''')
        print ('-' * 40)
        print ('')


    else:
        total_spent < 20000
        loyalty_points = total_spent / 100
        print ('Loyalty Points:',loyalty_points)
        print ('Customer Status: Standard')
        print ('-' * 40)
        print (f'''Thank You For Your Patronage!
We hope to See You Again.''')
        print ('-' * 40)
        print ('')


calculate_account_status ()
calculate_account_status ()
calculate_account_status ()
calculate_account_status ()
```
