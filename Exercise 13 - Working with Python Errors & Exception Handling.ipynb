## Ex1: Hospital Patient Data Validator

```python
# Task : build a small Patient Data Validator that can identify and handle these problems without unexpectedly crashing.

# I intend to solve this task by using a self-built function called 'Validate_patient_data ()', with while loops regulating the number of attempts...
# ... the user has.

def validate_patient_data ():
    """patient_name, patient_ID, age and the weight values are accepted using the input() function."""
    patient_name = input ('Patient Name:').strip().title()
    patient_ID = input ('Patient ID:').strip().upper()
    '''patient_name and patient_ID are formatted using the [.strip(), .title(), and .upper()] methods'''
    age_attempts = 0
    age_valid = False
    while age_attempts < 3:
        try:
            age = int (input('Age:'))
            age_attempts += 1

            if age > 0:
                age_valid = True
                print ('Age Accepted')
                break
            else:
                age <= 0
                print ('Invalid Age!, Age cannot be Zero or Negative')

        except ValueError:
            age_attempts += 1
            print ('Invalid Age! Patient\'s Age must be a Number.')
       
    if age_valid:

        weight_attempts = 0
        weight_valid = False

        while weight_attempts < 3:
            
            try:
                weight = float (input ('Weight:'))
                weight_attempts += 1

                if weight > 0:
                    print ('Weight Accepted')
                    weight_valid = True
                    break

                else:
                    print ('Invalid Weight!, Weight Must be greater than Zero')
                    
            except ValueError:
                print ('Invalid Weight!,Weight must be a Number')
                weight_attempts += 1

        if weight_valid:
            print (f'''
    ================ PATIENT RECORD ===================
    Patient Name: {patient_name}
    Patient ID: {patient_ID}
    Age: {age}
    Weight: {weight} kg
    Status: Validated
    ===================================================
    ''')

        else:
            print ('Maximum Weight-Input Attempts Reached, Please try again in 24hrs.')

    else:
        print ('Maximum Age-Input Attempts Reached!, Please try again in 24hrs.')


validate_patient_data ()
```

## Ex2: E-commerce Processing System

```python
# the task is to build a small Order Processing System

def process_order():
    customer_name = input ('Customer Name: ').strip().title()
    product_name = input ('Product Name: ').strip().title()

    price_attempts = 0
    price_valid = False

    while price_attempts < 3:
        try:
            product_price = int (input ('Price:'))
            price_attempts += 1

            if product_price > 0:
                price_valid = True
                print("Price Accepted.")
                break

            else:
                print("Invalid Price! Price must be greater than Zero.")

        except ValueError:
            price_attempts += 1
            print("Invalid Price! price must be a Number.")

    if price_valid:

        quantity_attempts = 0
        quantity_valid = False

        while quantity_attempts < 3:
            try:
                quantity = float(input ('Quantity:'))
                quantity_attempts += 1

                if quantity > 0:
                    quantity_valid = True
                    print("Quantity Accepted.")
                    break

                else:
                    print("Invalid Quantity! Quantity must be greater than Zero.")

            except ValueError:
                quantity_attempts += 1
                print("Invalid Quantity! Quantity must be a Number.")


        if quantity_valid:
            total_cost = product_price * quantity
            print (f'''
========== ORDER SUMMARY ==========
Customer: {customer_name}
Product: {product_name}
Unit Price: #{product_price}
Quantity: {quantity}
Total Cost: #{total_cost}
Order Status: Approved
===================================
''')

        else:
            print ('Maximum attempts to input quantity exhausted!')

    else:
        print ('Maximum Attempts to input Price Exhausted!')
            

process_order ()
```

## Ex3: Banking Transaction Validator

```python
# task is to build a Banking Transaction Validator.

def validate_transaction ():
    balance = 150000

    withdrawal_attempts = 0
    withdrawal_valid = False
# a while loop is used so that system can regulate the number of attempts the customer has to enter their withdrawal amount.
    while withdrawal_attempts < 5:
        try:
            withdrawal_amount = float (input('Withdrawal Amount:'))
            withdrawal_attempts += 1
# if the withdrawal is less than the balance of 150000, the systems accepts it.
            if 1 <= withdrawal_amount <= 150000  :
                withdrawal_valid = True
                print("Amount Accepted.")
                break
# withdrawal amount is higher than the balance of 150000, the system rejects it and displays 'Insufficient Funds'.
            elif withdrawal_amount > 150000:
                print("Insufficient Funds!")
# withdrawal amount is zero, the system rejects it and displays ' amount cannot be zero'
            elif withdrawal_amount == 0:
                print ('Inavlid! Amount cannot be Zero')
# withdrawal amount is negative, system rejects it and displays 'amount cannot be negative.'
            else:
                withdrawal_amount < 0
                print ('Invalid! Amount cannot be Negative')
# if withdrawal amount is not a number, system rejects it and dispplays 'amount must be a number.'
        except ValueError:
            withdrawal_attempts += 1
            print("Invalid! Amount must be a Number.")
# the transaction is processed below after the withdrawal amount has been validted
    if withdrawal_valid:
        remaining_balance = balance - withdrawal_amount

        print (f'''
========== TRANSACTION ==================
Account Balance: #{balance}
Withdrawal: #{withdrawal_amount}
Transaction Status: Successful
Remaining Balance: #{remaining_balance}
=========================================
''')


validate_transaction()
    
```
