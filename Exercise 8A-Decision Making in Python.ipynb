## Ex2: Employee Performance Rating

```python
# the code below is to create a varibale for employee performance score rating
performance_score = int(input('Employee\'s Score: ',))

# the code below is to determine an employee's performance rating based on their score
if performance_score >= 90:
    print ('Outstanding')

elif 75 <= performance_score <= 89:
    print ('Exceeds Expectations')

elif 60 <= performance_score <= 74:
    print ('Meets Expectation')

else :
    performance_score < 60
    print ('Needs Improvement')


    
```
