## Mini Project: Secure Employee Access Verification System

```python
# the code below is to create variables for employees' security requirements to grant access.
employee_name = input ('Employee Name: ',)
age = int(input ('Age: ',))
employment_status = input ('Employment Status: ',)
department = input ('Department: ',)
security_clearance_level = int(input ('Clearance Level: ',))


# The employee must satisfy all of the following conditions:
# Be 18 years or older
# Be an Active employee
# Belong to either the IT department ,the Cybersecurity department or the Executive department
# Have a security clearance level of 3 or above


# the code below is to verify all security requirements using the AND logical operator
# and also to determine if access would be granted to the employee or not.

if age >= 18 and employment_status == 'Active' and (department == 'Cybersecurity' or department == 'IT' or department == 'Executive') and security_clearance_level >= 3 and security_clearance_level == 3:
    print (f'''Access Granted
Security Level: Standard Secure Access
    ''')

elif age >= 18 and employment_status == 'Active' and ( department == 'Cybersecurity' or department == 'IT' or department == 'Executive') and security_clearance_level >= 3 and security_clearance_level == 4:
    print (f'''Access Granted
Security Level: Advanced Secure Access
    ''')

elif age >= 18 and employment_status == 'Active' and (department == 'Cybersecurity' or department == 'IT' or department == 'Executive') and security_clearance_level >= 3 and security_clearance_level >= 5 :
    print (f'''Access Granted
Security Level: Top Secret Access
    ''')

else:
    print ('Access Denied')

    
```
