# Ex1: Daily Patient Temperature Monitor (Health Care)

```python
# the code below is to Use a for loop to displays patient numbers from 1 to 10.

for patient in range (1, 11):
    print (f'Checking Patient {patient}')

print (f'''
Morning Temperature Check Completed.
''')

# the code below is to print only even numbers of the patients

print (f'''
======================================================================================
                BONUS CHALLENGE

''')

for patient in range (1, 11):
    if patient % 2 == 0:
        print (f'Checking Patient {patient}')

print (f'''
Morning Temperature Check for Patients on Even Numbers Completed.
''')
```

# Ex2: ATM LOGIN SECURITY SYSTEM (BANKING)

```python
# the code below is for a program that checks the successful login of customers on the mobile app.
# if customer login is successful, system should stop asking for number of attempts

successful = True

for attempt in range (1,4):
    if successful:
        print ('Attempt', attempt)
        break

print ('Login Successful')

print (f'''

-----------------------------------------------------------------------------------------------
            BONUS CHALLENGE

''')

# the code below is to display the number of failed attempts to login if the customers fails to login after three attempts.

successful = False

for attempt in range (1,4):
    if successful:
        print ('Attempt', attempt)
        break
else:
    print (f'''Login Failed,
Account Locked.
    ''')
```

# Ex3: WAREHOUSE INVENTORY SCANNER (RETAIL)

```python
# the code below is to display the scanning process of different products in a retail shop using the for loop statement.
products = [
    'Laptop',
    'Keyboard',
    'Mouse',
    'Printer',
    'Monitor'
]

for product in products:
    print (f'Scanning {product}')

print (f'''
Inventory Scan Completed Successfully
''')

print (f'''
------------------------------------------------------------------------------------------------
BONUS CHALLENGE : using Range () and len ()

''')

for i in range(len(products)):
    print (f'Product {i + 1} : {products[i]}')
        

print (f'''
------------------------------------------------------------------------------------------------
BONUS CHALLENGE : using enumerate ()

''')

for number, product in enumerate (products, start =1):
    print (f'Product {number} : {product}')
```

# Ex5: Employee Payroll Verification (Human Resources)

```python
# the code below is to display even-numbered employee IDs in a payroll batch, using the count, range () and step functions.
count = 0

for employee_Id in range (1,21):
    if employee_Id % 2 == 0:
        count += 1
        print (f'Processing Employee {employee_Id} - Eligible')
print('------------------------------------------------------------')
print (f'''
Total Employees Processed: {count} 
''')
```

# Mini Project : Hospital Monitoring System

```python
# the code below is to display a monitoring system for LifeCare Hospital.

print (f'''
========================================================================
        LIFECARE HOSPITAL MONITORING SYSTEM
========================================================================

''')
count = 0
count2 = 0
count3 = 0


for patient in range(1,11):
    print (f'Checking Patient {patient}')
    count += 1
    if patient % 2 == 0:
        count2 += 1
        print (f'''
Patient {patient} - Blood Pressure Checked
        ''')
        print ('------------------------------------------')
    if patient % 5 == 0:
        count3 += 1
        print (f'''
Patient {patient} - Critical Patient.
''')
        print ('------------------------------------------')
print (f'Total Blood Pressure Checks: {count2}')

print (f'''
========================================================================
        BONUS CHALLENGE

''')

for patient in range (1,11):
    print (f'''
Checking Patient {patient} Temperature, Pulse & Blood Pressure...
''')
    if patient % 5 == 0:
        count3 += 1
        print (f'''
Patient {patient} - Critical Patient.
''')
        print('------------------------------------------------------------')
        print ('Critical Patient Found')
        print('------------------------------------------------------------')
        break
        
print (f'Total Patients Checked: {count}')
print (f'Total Blood Pressure Checks: {count2}')
print (f'''
Number of Critical Patients Detected: {count3 -1}
''')
# there is a (minus 1) in the code above {count3 -1} to avoid the system recording duplicate number of critical patient detected
print('-------------------------------------------------------------------')

print (f'''
Daily Patient Monitoring Completed.
''')
```
