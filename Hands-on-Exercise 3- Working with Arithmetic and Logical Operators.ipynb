## Question 4: Banking Withdrawal Checker

```python
account_balance = 520000
withdrawal_amount = 150000
remaining_balance = (account_balance - withdrawal_amount)
print ("Remaining Balance:" , remaining_balance)
print ("Withdrawal Approved:" , account_balance >= withdrawal_amount)
```

## Bonus Challenge : Smart Shopping Discount Calculator

```python
# This line is a comment to notify you that the items listed below are customer purchased Items
laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000
total_purchased_amount = (laptop_bag + wireless_mouse + keyboard + usb_drive)
print ("Toatal Purchased Amount:" , total_purchased_amount)
print ("Customer Qualifies for Discount:" ,total_purchased_amount >= 50000)
discount_amount = ((total_purchased_amount // 100) * 10)
final_amount_to_pay = (total_purchased_amount - discount_amount)
print ("Discount Amount:" , discount_amount)
print ("Final Amount To Pay:", final_amount_to_pay)
```
