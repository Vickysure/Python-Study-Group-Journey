## Question 3: Inventory Management System (Mini Project)

```python
# this is a code for inventory management system
stock = 150
sold = 28
arrivals = 75
item_per_carton = 12
stock -= sold
stock += arrivals
print ("Updated Stock:", stock)
stock //= item_per_carton
print ("Number of Full Cartons:", stock)
```

```python
# this is a code for inventory management system
stock = int (input ("enter Initial stock:"))
sold = int (input("enter sold items:"))
arrivals = int (input("enter new arrivals:"))
item_per_carton = int (input ("how many items per carton?:"))
stock -= sold
print ("Remaining Stock after sales:", stock)
stock += arrivals
print ("Updated Stock after Restocking:", stock)
total_stock = stock
stock %= item_per_carton
print ("Remaining Loose Items:", stock)
total_stock //= item_per_carton
print ("Number of Full Cartons:", total_stock)
```
