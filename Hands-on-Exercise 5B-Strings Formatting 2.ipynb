## Ex2: Payroll System

```python
#the code below is to create variables for an employee receipt
employee_name = 'David Musa'
department = 'Finance'
basic_salary = 185000
bonus = 35000
tax_rate = 7.5
task = 'PAYROLL SUMMARY'
divider = '====================================================='
space ='                                                 '

# the code below is to calculate gross salary of the employee
gross_salary = basic_salary + bonus

#the code below is to calculate the tax amount of the employee
tax_amount = (gross_salary / 100) * tax_rate

#the code below is to calculate the net salary of the employee
net_salary = (gross_salary - tax_amount) 

#the code below is to assign a line for the payroll report construction

line1 = f'{divider}'
line2 = f'      {task}'
line3 = f'{divider}'
line4 = f'{space}'
line5 = f'Employee: {employee_name}'
line6 = f'Department: {department}'
line7 = f'Basic Salary: #{basic_salary:.2f}'
line8 = f'Bonus: #{bonus:.2f}'
line9 = f'Gross Salary: #{gross_salary:.2f}'
line10 = f'Tax Amount(7.5%): #{tax_amount:.2f}'
line11 = f'Net Salary: #{net_salary:.2f}'
line12 = f'{space}'
line13 = f'{divider}'

# the code below is to print all of the payroll report construction lines

print (line1)
print (line2)
print (line3)
print (line4)
print (line5)
print (line6)
print (line7)
print (line8)
print (line9)
print (line10)
print (line11)
print (line12)
print (line13)
```
