## Ex1: Hospital Patient Registration System

```python
# task: Organize patient registration using data structures
# the code below is to create a list containing patient names
patients = ['James','Sarah','David','Grace','Esther']

# the code below is to display the entire list using the print function
print (f'''Patient Names:
{patients}
''')

# the code below is to display the total number of patient's names using the len() function
print (f'''Total Number of Patients:',
{ len(patients)}
''')

# the code below is to add a new patient to the list using the (.append) method
patients.append('Michael')

# the code below is to display the updated list
print (f'''Updated Patients List:
{patients}
''')
```

## Ex2: University Examination Records

```python
# the code below is to create a tuple named months for the months of the year
months = ('January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December')

# the code below is to diplay the tuple named months
print (f'''Months:
{months}
''')

# the code below is to display the number of months in the tuple
print (f'''Number of Months:
{len(months)}
''')

# using a tuple for this program is a better choice because we need the number of months to be constant and not change.
# since tuples do not allow modifications after creation unlike list, it is then a better choice to use tuple instead of list
```

## Ex3: Customer Loyalty Program

```python
# the code is to create a set for customer IDs
customer_ID = {101,205,101,310,450,205,512,310,620}


# the code below is to display the customer_IDs
print (f'''Customer IDs:
{customer_ID}
''')

# the code below is to display the total number of unique customers
print (f'''Total Number of Unique Customers:
{len(customer_ID)}
''')

# the duplicate customer IDs disappeared because a set only stores unique values and does not allow duplicates.
```

## Ex5: Business Sales Performance Dashboard

```python
# the code below is to create a list named sales
sales = [2500,3200,4100,2800,3900,3500]

# the code below is to display the sales list
print (f'''Sales Figures:
{sales}
''')

# the code below is to display the total number of branches
print (f'''Total Number of Branches:
{len(sales)}
''')


# the code below is to display highest sales from the list
print (f'''Highest Sales Figure:
{max(sales)}
''')

# the code below is to display the lowest sales figure from the list
print (f'''Lowest Sales Figure:
{min(sales)}
''')

# the code below is to display total sales
print (f'''Total Sales:
{sum(sales)}
''')

# the code is to add a new branch with sales 4700
sales.append(4700)

# the code below is to display the updated list
print (f'''Updated Sales List:
{sales}
''')

# the code below is to display the new total number of branches
print (f'''New Total Number of Branches:
{len(sales)}
''')


```
